```python
df['sdate'] = pd.to_datetime(df['Ship Date'], dayfirst = True)
df['odate'] = pd.to_datetime(df['Order Date'] , dayfirst = True)
```

```python
df['omonth'] = df['odate'].dt.month
df['smonth'] = df['sdate'].dt.month
```

```python
df['oyear'] = df['odate'].dt.year
df['syear'] = df['sdate'].dt.year
```

```python
df['oquarter'] = df['odate'].dt.quarter
df['squarter'] = df['sdate'].dt.quarter
```

```python
df.drop(columns = ['Order Date' , 'Ship Date'] , axis = 1 , inplace = True)
```

```python
df = df.dropna(subset = ["Postal Code"])
```

```python
# Outliers Detection
q3 = df['Sales'].quantile(0.75)
q1 = df['Sales'].quantile(0.25)
iqr = q3 - q1
```

```python
high = q3 + 1.5 * iqr
low = q1 - 1.5 * iqr
```

```python
# Outliers
df[(df['Sales'] > high) | (df['Sales'] < low)]
```

```python
print(df.columns.tolist())
```

```python
df["Shipping Days"] = (pd.to_datetime(df["sdate"]) - pd.to_datetime(df["odate"])).dt.days
```

```python
df.drop(columns = ['sdate' , 'odate'] , axis = 1 , inplace = True)
```
